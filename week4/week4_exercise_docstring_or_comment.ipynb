## Adding Docstring or Comment in the code

```python
import os
import io
import sys
import anthropic
```

```python
os.environ['ANTHROPIC_API_KEY'] = os.getenv('ANTHROPIC_API_KEY', '*****')
```

```python
CLAUDE_MODEL = "claude-3-5-sonnet-20240620"
claude = anthropic.Anthropic()
```

```python
system_message = "You are an assistant who will add Docstring or comment in the code "
system_message += "so that people can easily understand the functionality of code snippet, add comment before each function or method about its functionality "
system_message += " and the parameters it is expecting and the return type."
```

```python
def user_prompt_for(code_snippet):
    user_prompt = "Add docstring and comment before each function or method\n\n"
    user_prompt += code_snippet
    return user_prompt
```

```python
def messages_for(code_snippet):
    return [
        {"role": "system", "content": system_message},
        {"role": "user", "content": user_prompt_for(code_snippet)}
    ]
```

```python
def generate_comment_claude(code_snippet):
    result = claude.messages.stream(
        model=CLAUDE_MODEL,
        max_tokens=2000,
        system=system_message,
        messages=[{"role": "user", "content": user_prompt_for(code_snippet)}],
    )
    reply = ""
    with result as stream:
        for text in stream.text_stream:
            reply += text
            print(text, end="", flush=True)
    #write_output(reply)
```

```python
code_snippet = """
import time

def calculate(iterations, param1, param2):
    result = 1.0
    for i in range(1, iterations+1):
        j = i * param1 - param2
        result -= (1/j)
        j = i * param1 + param2
        result += (1/j)
    return result

start_time = time.time()
result = calculate(100_000_000, 4, 1) * 4
end_time = time.time()

print(f"Result: {result:.12f}")
print(f"Execution Time: {(end_time - start_time):.6f} seconds")
"""
```

```python
generate_comment_claude(code_snippet)
```
